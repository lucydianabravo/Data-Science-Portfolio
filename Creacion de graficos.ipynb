```python
import pandas as pd
df = pd.read_csv('StudentsPerformance.csv')
df
```

```python
#Creación de grafico de stripplot (grafico de bandas)
import seaborn as sns
sns.stripplot(data=df, y="writing score", x="math score", jitter=True)
```

```python
import matplotlib.pyplot as plt
fig, ax= plt.subplots()
sns.stripplot(data=df, y="writing score", x="math score", jitter=True)
ax.axes.xaxis.set_ticks([])
plt.show()
```

```python
import matplotlib.pyplot as plt
fig, ax= plt.subplots()
sns.stripplot(data=df, y="writing score", x="race/ethnicity", jitter=True)
ax.axes.xaxis.set_ticks([])
plt.show()
```

```python
#Creación de swarmplot (Driagrama de enjambre)
sns.swarmplot(data=df, y="writing score", x="race/ethnicity")
```

```python
#Creación de boxplot . Diagrama de caja
sns.boxplot(data=df, y="writing score", x="race/ethnicity")
```

```python
#Creación de grafico de violin
sns.violinplot(data=df, y="writing score", x="race/ethnicity")
```

```python
#Creación de lvplot (letter value plot)
sns.catplot(data=df, y="writing score", x="race/ethnicity", kind="boxen")
```

```python
#Creación de barplot
sns.barplot(data=df, y="writing score", x="race/ethnicity")
```

```python
#Creación de barplot
sns.barplot(data=df, y="writing score", x="race/ethnicity", hue="lunch") #lunch es la vbariable que quiero
```

```python
#Creación de pointplot
sns.pointplot(data=df, y="writing score", x="race/ethnicity", hue="lunch")
```

```python
#Creación de grafico de regresion
sns.regplot(data=df, y="writing score", x="race/ethnicity", marker= "+")
```

```python
#Creación de countplot
sns.countplot(data=df, y="writing score", hue="lunch")
```

```python
#Grafico de residuales
sns.residplot(data=df, y="writing score", x="race/ethnicity")
```

```python
#Graficos de regresion con regplot y media por valor de x
#Muy util para enfatizar tendencias
import numpy as np
sns.regplot(data=df, y="writing score", x="race/ethnicity", x_estimator= np.mean)
```
