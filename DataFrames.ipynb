```python
import pandas as pd
import numpy as np
arregloperros = np.array([['Comandante Pipo', 'Terry Mix', 'negro', 45, 14, '6 de mayo'],
        ['Chipote', 'Pastor tobillero con ratonero chihuahua y pug', 'negro', 35, 9, '31 de octubre'],
        ['Lenteja', 'Mestizo', 'negro', 35, 10, '5 de noviembre']])
arregloperros
```

```python
df = pd.DataFrame(data =arregloperros, columns= ['Nombre', 'Raza', 'Color', 'Estatura hasta la cruz', 'Peso', 'Cumpleaños'])
df
```

```python
df['Estatura hasta la cruz'] = df['Estatura hasta la cruz'].astype(int)
```

```python
df[df['Estatura hasta la cruz']<50]
```

```python
df['Peso'] = df['Peso'].astype(int)
```

```python
df['IMM'] = df['Peso']/(df['Estatura hasta la cruz']/100)**2
df
```
